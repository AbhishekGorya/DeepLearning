```python
X = []
y = []

for seq in sequence:
  for i in range(1,len(seq)):
    input_seq = seq[:i]
    output_seq = seq[i]
    X.append(input_seq)
    y.append(output_seq)
```

```python
#lets create first simple rnn model
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN,LSTM,  Dense
```

```python
#lets create 50 dimentional embeddings
```

```python
embedding_dim = 50
```

```python
rnn_units = 128 #hidden neuron
```

```python
rnn_model = Sequential()
```

```python
#lets create 3 dimentional embedding
```

```python
rnn_model.add(Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_len))
#lets create simple rnn layers
rnn_model.add(SimpleRNN(units=rnn_units, activation='tanh', return_sequences=True)) #hidden
rnn_model.add(Dense(units=vocab_size, activation='softmax'))   #output
```

```python
rnn_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
index_to_word = {}
for word , index in word_index.items():
  index_to_word[index] = word
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
def predictor(model, tokenizer, text, max_len):

    text = text.lower()

    seq = tokenizer.texts_to_sequences([text])[0]

    seq = pad_sequences(
        [seq],
        maxlen=max_len,
        padding='pre'
    )

    pred = model.predict(seq, verbose=0)

    pred_index = np.argmax(pred, axis=-1)[0]

    print("Predicted Index:", pred_index)

    return index_to_word.get(pred_index, "")
```

```python
seed_text = "what are you"
next_word = predictor(lstm_model,tokenizer,seed_text,max_len)
print(next_word)
```

```python
def generate_text(model,tokenizer,seed_text,max_len,n_words):
  for _ in range(n_words):
    next_word = predictor(model,tokenizer,seed_text,max_len)
    if next_word == "":
      break
    seed_text += " " + next_word
  return seed_text
```
